```python
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
API_KEY = os.getenv("GEMINI_API_KEY")

# Modelo que quieres usar
MODEL = "gemini-2.5-flash-lite"

# API key de Gemini
# API_KEY = userdata.get('GEMINI_API_KEY')
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate, SystemMessagePromptTemplate, HumanMessagePromptTemplate
from langchain_core.messages import HumanMessage, AIMessage

from pydantic import BaseModel, Field

from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
# Load PDF
ruta = "../data/Montaje_de_un_Motor_Eléctrico_Básico.pdf"
loader = PyPDFLoader(ruta)
pages = loader.load()
```

```python
# Dividir el texto en chunks mas pequeños
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=2000,
    chunk_overlap=100
)

chunks = text_splitter.split_documents(pages)
```

```python
# 3. Pasar el texto al LLM
llm = ChatGoogleGenerativeAI(model=MODEL, temperature=0.4, google_api_key=API_KEY)

resumenes = []

for chunk in chunks:
    response = llm.invoke(f"Haz un resumen de los puntos mas importantes del siguiente text: {chunk.page_content}")
    resumenes.append(response.content)

resumen_final = llm.invoke(f"Combina y junta en un solo resumen los siguientes puntos en un resumen coherente y final: {" ".join(resumenes)}")
```
