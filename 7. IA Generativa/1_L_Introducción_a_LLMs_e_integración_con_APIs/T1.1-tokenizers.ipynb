```python
# 3. Tipos de tokenización (visión general)
# 3.1 Word-based (obsoleto)

# Cada palabra = 1 token

# Problemas con vocabulario infinito

# 3.2 Character-based

# Cada carácter = 1 token

# Demasiados tokens

# 3.3 Subword-based ✅ (estándar actual)

# Aprenden fragmentos frecuentes de palabras.

# Algoritmos más usados:

# BPE (Byte Pair Encoding) – GPT

# WordPiece – BERT

# Unigram LM – SentencePiece
```

```python
import tiktoken

enc = tiktoken.encoding_for_model("gpt-4o-mini")
text = "La inteligencia artificial está cambiando el mundo"
text1 = "Me siento en un banco"
text2 = "Voy al banco a sacar dinero"
tokens = enc.encode(text)
```

```python
print("Número de tokens:", len(tokens))
print("IDs:", tokens)
print("Tokens decodificados:")
print([enc.decode([t]) for t in tokens])

tokens1 = enc.encode(text1)

print("Número de tokens:", len(tokens1))
print("IDs:", tokens1)
print("Tokens decodificados:")
print([enc.decode([t]) for t in tokens1])

tokens2 = enc.encode(text2)

print("Número de tokens:", len(tokens2))
print("IDs:", tokens2)
print("Tokens decodificados:")
print([enc.decode([t]) for t in tokens2])
```

```python
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace
```

```python
text = "Tokenización en IA generativa es clave"

# Tokenizador BPE simple
bpe_tokenizer = Tokenizer(BPE())
bpe_tokenizer.pre_tokenizer = Whitespace()

output = bpe_tokenizer.encode(text)
print("Tokens BPE:", output.tokens)
print("Número de tokens:", len(output.tokens))
```

```python
tokenizer = Tokenizer.from_pretrained("bert-base-uncased")

output = tokenizer.encode("Me siento en un banco")
print(output.tokens)
```

```python
with open("../data/data_01/dataset_entrenamiento_lstm.txt", encoding="latin-1") as f:
    text = f.read().splitlines()
```

```python
text[:10]
```

```python
import string
import unicodedata

def preproc_texto(txt):
    txt = "".join(c for c in txt if c not in string.punctuation).lower()
    txt = unicodedata.normalize('NFKD', txt).encode('ascii', 'ignore').decode('utf-8', 'ignore')
    return txt
```

```python
dataset = [preproc_texto(t) for t in text]
```

```python
dataset[:10]
```

```python
trainer = BpeTrainer(
    vocab_size=100,                # pequeño a propósito
    min_frequency=2,               # ignora tokens raros
    special_tokens=["[UNK]"]
)

bpe_tokenizer.train_from_iterator(dataset, trainer)
```

```python
# Context window
# Los modelos tienen un límite de tokens por entrada + salida.

long_text = "Esto es una frase. " * 500
print(len(enc.encode(long_text)))

# 💰 Más tokens = más coste
# 🧠 Más tokens = menos contexto útil
# ⚠️ Prompt largo ≠ prompt mejor
```
